```python
import pandas as pd

df = pd.read_csv('../data/311_raw.csv')
pd.set_option('display.max_columns', None)

df['open_date'] = pd.to_datetime(df['open_date'], format='ISO8601')
df['close_date'] = pd.to_datetime(df['close_date'], format='ISO8601')
df['target_close_date'] = pd.to_datetime(df['target_close_date'], format='ISO8601')
```

```python
df['response_time_hours'] = (df['close_date'] - df['open_date']).dt.total_seconds() / 3600
df['response_time_hours'].describe()
```

```python
df[df['response_time_hours'] > 720]['case_topic'].value_counts()  # cases over 30 days
```

```python
df.groupby('neighborhood')['response_time_hours'].agg(['median', 'count']).sort_values('median')
```
